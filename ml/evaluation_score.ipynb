```python
import pandas as pd
from IPython.display import display, HTML

df_eval   = pd.read_csv('output/evaluation_v3.csv')
df_tuning = pd.read_csv('output/tuning_v3.csv')
df_produk = pd.read_csv('output/evaluation_per_produk_v3.csv')
```

```python
display(HTML("<h3>Tabel 1 Perbandingan Model (Test Set: Juli–Agustus 2026)</h3>"))

df_cmp = df_eval[df_eval['model'].isin(['XGBoost tuned', 'Holt-Winters'])].copy()
df_cmp.columns = ['Model', 'MAE', 'RMSE', 'MAPE (%)', 'WAPE (%)']
df_cmp[['MAE','RMSE','MAPE (%)','WAPE (%)']] = \
    df_cmp[['MAE','RMSE','MAPE (%)','WAPE (%)']].round(3)

def highlight_min_dark(s):
    is_min = s == s.min()
    return ['background-color: #2d6a4f; color: white; font-weight: bold'
            if v else '' for v in is_min]

df_cmp.reset_index(drop=True).style \
    .apply(highlight_min_dark, subset=['MAE','RMSE','MAPE (%)','WAPE (%)']) \
    .set_caption("Hijau = nilai terbaik per metrik")
```

```python
display(HTML("<h3>Tabel 2 Hyperparameter Tuning XGBoost (Validation Set)</h3>"))

df_t = df_tuning[df_tuning['candidate'].isin(['1','2','3','Holt-Winters'])].copy()
df_t = df_t[['candidate','MAE','RMSE','MAPE','WAPE']]
df_t.columns = ['Kandidat','MAE','RMSE','MAPE (%)','WAPE (%)']
df_t[['MAE','RMSE','MAPE (%)','WAPE (%)']] = \
    df_t[['MAE','RMSE','MAPE (%)','WAPE (%)']].round(3)

df_t.reset_index(drop=True).style \
    .apply(highlight_min_dark, subset=['MAE','RMSE','MAPE (%)','WAPE (%)']) \
    .set_caption("Hijau = nilai terbaik per metrik")
```

```python
display(HTML("<h3>Tabel 3 Evaluasi MAE per Produk (Validation Set)</h3>"))

df_p = df_produk[['produk_id','xgb_MAE','holt_MAE']].copy()
df_p.columns = ['Produk ID', 'XGBoost MAE', 'Holt-Winters MAE']
df_p['XGBoost MAE']      = df_p['XGBoost MAE'].round(3)
df_p['Holt-Winters MAE'] = df_p['Holt-Winters MAE'].round(3)

def highlight_min_row(row):
    vals = row[['XGBoost MAE','Holt-Winters MAE']]
    return [
        'background-color: #2d6a4f; color: white; font-weight: bold' if v == vals.min() else ''
        for v in row
    ]

xgb_wins  = (df_p['XGBoost MAE'] < df_p['Holt-Winters MAE']).sum()
holt_wins = (df_p['Holt-Winters MAE'] < df_p['XGBoost MAE']).sum()

df_p.reset_index(drop=True).style \
    .apply(highlight_min_row, axis=1, subset=['XGBoost MAE','Holt-Winters MAE']) \
    .set_caption(f"Hijau = MAE lebih kecil per baris | XGBoost unggul: {xgb_wins} | Holt-Winters unggul: {holt_wins}")
```
