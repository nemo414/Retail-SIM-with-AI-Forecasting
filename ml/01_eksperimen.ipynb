```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. LOAD & PREPARE DATA (Memastikan 'qty' tersedia)
# ---------------------------------------------------------
path_csv = '../data/agregasi_harian.csv'
df_raw = pd.read_csv(path_csv)
df_raw.columns = df_raw.columns.str.strip().str.lower()
df_raw['tanggal'] = pd.to_datetime(df_raw['tanggal'])

# Penyesuaian nama kolom jika diperlukan
if 'qty' not in df_raw.columns:
    col_qty = [c for c in df_raw.columns if c not in ['tanggal', 'produk_id']][0]
    df_raw = df_raw.rename(columns={col_qty: 'qty'})

# Buat grid lengkap (365 hari x 20 produk)
semua_tanggal = pd.date_range(start=df_raw['tanggal'].min(), end=df_raw['tanggal'].max(), freq='D')
semua_produk = df_raw['produk_id'].unique()
grid = pd.MultiIndex.from_product([semua_tanggal, semua_produk], names=['tanggal', 'produk_id']).to_frame().reset_index(drop=True)

df = pd.merge(grid, df_raw[['tanggal', 'produk_id', 'qty']], on=['tanggal', 'produk_id'], how='left')
df['qty'] = df['qty'].fillna(0)

# ---------------------------------------------------------
# 2. JOB 2: EDA & SIMPAN PLOT
# ---------------------------------------------------------
os.makedirs('../docs/ml', exist_ok=True)

plt.figure(figsize=(12, 4))
total_harian = df.groupby('tanggal')['qty'].sum()
plt.plot(total_harian.index, total_harian.values, color='#1f77b4', linewidth=1.5)
plt.title('Pola Permintaan Harian Ritel (Sep 2025 - Agu 2026)')
plt.xlabel('Tanggal')
plt.ylabel('Total Qty Terjual')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('../docs/ml/eda_pola_harian.png')
plt.show()

# ---------------------------------------------------------
# 3. JOB 5: REKAYASA FITUR (FEATURE ENGINEERING)
# ---------------------------------------------------------
df = df.sort_values(['produk_id', 'tanggal']).reset_index(drop=True)

# Fitur Tanggal & Kalender
df['day_of_week'] = df['tanggal'].dt.dayofweek
df['is_weekend'] = df['day_of_week'].isin([5, 6]).astype(int)
df['day_of_month'] = df['tanggal'].dt.day
df['is_awal_bulan'] = (df['day_of_month'] <= 5).astype(int)
df['month'] = df['tanggal'].dt.month

# Fitur Lag (1, 7, 14 hari)
df['lag_1'] = df.groupby('produk_id')['qty'].shift(1)
df['lag_7'] = df.groupby('produk_id')['qty'].shift(7)
df['lag_14'] = df.groupby('produk_id')['qty'].shift(14)

# Fitur Rolling Mean (7 & 28 hari)
df['rolling_mean_7'] = df.groupby('produk_id')['qty'].transform(lambda x: x.shift(1).rolling(7).mean())
df['rolling_mean_28'] = df.groupby('produk_id')['qty'].transform(lambda x: x.shift(1).rolling(28).mean())

# Hapus baris NaN akibat pembuatan lag/rolling mean
df_features = df.dropna().reset_index(drop=True)

print(f"Ukuran Data Fitur: {df_features.shape}")
df_features.head()
```

```python
import warnings
import os
import pandas as pd
import numpy as np
from statsmodels.tsa.api import ExponentialSmoothing

warnings.filterwarnings('ignore')

# ---------------------------------------------------------
# 0. PEMERIKSAAN MANDIRI (MENCEGAH NAMEERROR)
# ---------------------------------------------------------
if 'df_features' not in locals():
    path_csv = '../data/agregasi_harian.csv'
    df_raw = pd.read_csv(path_csv)
    df_raw.columns = df_raw.columns.str.strip().str.lower()
    df_raw['tanggal'] = pd.to_datetime(df_raw['tanggal'])

    if 'qty' not in df_raw.columns:
        col_qty = [c for c in df_raw.columns if c not in ['tanggal', 'produk_id']][0]
        df_raw = df_raw.rename(columns={col_qty: 'qty'})

    semua_tanggal = pd.date_range(start=df_raw['tanggal'].min(), end=df_raw['tanggal'].max(), freq='D')
    semua_produk = df_raw['produk_id'].unique()
    grid = pd.MultiIndex.from_product([semua_tanggal, semua_produk], names=['tanggal', 'produk_id']).to_frame().reset_index(drop=True)

    df = pd.merge(grid, df_raw[['tanggal', 'produk_id', 'qty']], on=['tanggal', 'produk_id'], how='left')
    df['qty'] = df['qty'].fillna(0)
    df = df.sort_values(['produk_id', 'tanggal']).reset_index(drop=True)

    df['day_of_week'] = df['tanggal'].dt.dayofweek
    df['is_weekend'] = df['day_of_week'].isin([5, 6]).astype(int)
    df['day_of_month'] = df['tanggal'].dt.day
    df['is_awal_bulan'] = (df['day_of_month'] <= 5).astype(int)
    df['month'] = df['tanggal'].dt.month

    df['lag_1'] = df.groupby('produk_id')['qty'].shift(1)
    df['lag_7'] = df.groupby('produk_id')['qty'].shift(7)
    df['lag_14'] = df.groupby('produk_id')['qty'].shift(14)

    df['rolling_mean_7'] = df.groupby('produk_id')['qty'].transform(lambda x: x.shift(1).rolling(7).mean())
    df['rolling_mean_28'] = df.groupby('produk_id')['qty'].transform(lambda x: x.shift(1).rolling(28).mean())

    df_features = df.dropna().reset_index(drop=True)

# ---------------------------------------------------------
# 1. JOB 3: SPLIT DATA BERDASARKAN WAKTU & METRIK EVALUASI
# ---------------------------------------------------------
# Train: Sep 2025 - Apr 2026 | Val: Mei - Jun 2026 | Test: Jul - Agu 2026
train = df_features[(df_features['tanggal'] >= '2025-09-01') & (df_features['tanggal'] <= '2026-04-30')].copy()
val   = df_features[(df_features['tanggal'] >= '2026-05-01') & (df_features['tanggal'] <= '2026-06-30')].copy()
test  = df_features[(df_features['tanggal'] >= '2026-07-01') & (df_features['tanggal'] <= '2026-08-31')].copy()

print(f"Jumlah baris -> Train: {len(train)}, Val: {len(val)}, Test: {len(test)}")

def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    mae = np.mean(np.abs(y_true - y_pred))
    rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
    
    mask = y_true > 0
    mape = np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100 if np.any(mask) else np.nan
    
    total_aktual = np.sum(y_true)
    wape = (np.sum(np.abs(y_true - y_pred)) / total_aktual) * 100 if total_aktual > 0 else np.nan
    
    return {
        'MAE': round(mae, 4),
        'RMSE': round(rmse, 4),
        'MAPE (%)': round(mape, 2),
        'WAPE (%)': round(wape, 2)
    }

# ---------------------------------------------------------
# 2. JOB 4: BASELINE HOLT-WINTERS PER PRODUK
# ---------------------------------------------------------
train_val = pd.concat([train, val]).sort_values(['produk_id', 'tanggal'])

results_hw = []
fallback_count = 0

for pid in df_features['produk_id'].unique():
    df_p_train = train_val[train_val['produk_id'] == pid].set_index('tanggal')['qty']
    df_p_test  = test[test['produk_id'] == pid].set_index('tanggal')['qty']
    
    try:
        model = ExponentialSmoothing(df_p_train, seasonal='add', seasonal_periods=7).fit()
        pred = model.forecast(len(df_p_test))
    except Exception:
        # Fallback jika konvergensi gagal
        pred = pd.Series(df_p_train.tail(7).mean(), index=df_p_test.index)
        fallback_count += 1
        
    metrics = calculate_metrics(df_p_test.values, pred.values)
    metrics['produk_id'] = pid
    results_hw.append(metrics)

df_res_hw = pd.DataFrame(results_hw)

print(f"\nJumlah Produk Fallback ke Moving Average: {fallback_count}")
print("\n--- Rata-Rata Metrik Baseline Holt-Winters (Test Set) ---")
print(df_res_hw[['MAE', 'RMSE', 'MAPE (%)', 'WAPE (%)']].mean())
```

```python
import os
import warnings
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import xgboost as xgb
from sklearn.model_selection import ParameterGrid

warnings.filterwarnings('ignore')

# ---------------------------------------------------------
# 1. JOB 6: FEATURE MATRIX & TUNING XGBOOST DI VALIDATION
# ---------------------------------------------------------
features = ['produk_id', 'day_of_week', 'is_weekend', 'day_of_month', 
            'is_awal_bulan', 'month', 'lag_1', 'lag_7', 'lag_14', 
            'rolling_mean_7', 'rolling_mean_28']
target = 'qty'

X_train, y_train = train[features], train[target]
X_val, y_val = val[features], val[target]
X_test, y_test = test[features], test[target]

# Grid Tuning pada Data Validation
param_grid = {
    'n_estimators': [300, 500],
    'max_depth': [4, 6],
    'learning_rate': [0.03, 0.05],
    'subsample': [0.8],
    'colsample_bytree': [0.8]
}

best_score = float('inf')
best_params = None

for p in ParameterGrid(param_grid):
    model_xgb = xgb.XGBRegressor(**p, random_state=42)
    model_xgb.fit(X_train, y_train)
    val_preds = model_xgb.predict(X_val)
    score = calculate_metrics(y_val, val_preds)['MAE']
    
    if score < best_score:
        best_score = score
        best_params = p

print(f"Parameter Terbaik (Validation): {best_params}")

# Latih ulang gabungan Train + Val dengan parameter terbaik
X_train_val = pd.concat([X_train, X_val])
y_train_val = pd.concat([y_train, y_val])

final_model = xgb.XGBRegressor(**best_params, random_state=42)
final_model.fit(X_train_val, y_train_val)

# Evaluasi SEKALI di Data Test
test_preds = final_model.predict(X_test)
test['xgb_pred'] = test_preds

xgb_metrics = calculate_metrics(y_test, test_preds)
print("\n--- Rata-Rata Metrik XGBoost (Test Set) ---")
print(xgb_metrics)

# ---------------------------------------------------------
# 2. SIMPAN GRAFIK AKTUAL VS PREDIKSI UNTUK PAPER/JURNAL
# ---------------------------------------------------------
os.makedirs('../docs/ml', exist_ok=True)
plt.figure(figsize=(12, 5))
for pid in [1, 2]:
    sample_prod = test[test['produk_id'] == pid].set_index('tanggal')
    plt.plot(sample_prod.index, sample_prod['qty'], label=f'Aktual Produk {pid}', alpha=0.7)
    plt.plot(sample_prod.index, sample_prod['xgb_pred'], linestyle='--', label=f'Prediksi XGBoost Produk {pid}')

plt.title('Perbandingan Penjualan Aktual vs Prediksi XGBoost (Test Set)')
plt.xlabel('Tanggal')
plt.ylabel('Qty Terjual')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('../docs/ml/aktual_vs_prediksi_xgboost.png')
plt.show()

# ---------------------------------------------------------
# 3. JOB 7: PERHITUNGAN SAFETY STOCK (SS) & REORDER POINT (ROP)
# ---------------------------------------------------------
stats_harian = train.groupby('produk_id')['qty'].agg(['mean', 'std']).reset_index()

Z = 1.65  # Service Level 95%
lead_time_hari = 3  # Asumsi lead time bawaan

stats_harian['safety_stock'] = np.ceil(Z * stats_harian['std'] * np.sqrt(lead_time_hari))
stats_harian['reorder_point'] = np.ceil((stats_harian['mean'] * lead_time_hari) + stats_harian['safety_stock'])

print("\n--- Contoh Tabel SS & ROP (5 Produk Pertama) ---")
print(stats_harian[['produk_id', 'mean', 'std', 'safety_stock', 'reorder_point']].head())
```
